### Problem Statement: Develop a knowledge base for a medical diagnosis system using First Order Logic predicates such as:
###   a. HasSymptom(Sankalp, Flu)
###   b. HasDisease(Kento, diarrhea) 


```python
facts = {("Fever", "Sankalp"), ("Cough", "Sankalp")}

rules = [
    ([("Fever", "Sankalp"), ("Cough", "Sankalp")], ("Flu", "Sankalp")),
    ([("Flu", "Sankalp")], ("NeedsRest", "Sankalp")),
    ([("SevereCough", "Sankalp"), ("ShortnessOfBreath", "Sankalp")], ("Pneumonia", "Sankalp"))
]

changed = True
while changed:
    changed = False
    for premises, conclusion in rules:
        if all(p in facts for p in premises) and conclusion not in facts:
            facts.add(conclusion)
            changed = True

queries = ["Flu", "NeedsRest", "Pneumonia"]
for disease in queries:
    if (disease, "Sankalp") in facts:
        print(f"{disease}(Sankalp) is True")
    else:
        print(f"{disease}(Sankalp) is False")
```
