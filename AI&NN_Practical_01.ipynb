```python
def a_star(grid, start, goal):
    def heuristic(a, b):
        return abs(a[0] - b[0]) + abs(a[1] - b[1])

    open_list = [start]
    came_from = {}
    g_score = {start: 0}
    f_score = {start: heuristic(start, goal)}

    while open_list:
        current = min(open_list, key=lambda x: f_score.get(x, float('inf')))
        if current == goal:
            path = []
            while current in came_from:
                path.append(current)
                current = came_from[current]
            path.append(start)
            return path[::-1]

        open_list.remove(current)

        for dx, dy in [(-1,0), (1,0), (0,-1), (0,1)]:
            neighbor = (current[0] + dx, current[1] + dy)
            if (0 <= neighbor[0] < len(grid)) and (0 <= neighbor[1] < len(grid[0])):
                if grid[neighbor[0]][neighbor[1]] == 1:
                    continue
                temp_g = g_score[current] + 1
                if temp_g < g_score.get(neighbor, float('inf')):
                    came_from[neighbor] = current
                    g_score[neighbor] = temp_g
                    f_score[neighbor] = temp_g + heuristic(neighbor, goal)
                    if neighbor not in open_list:
                        open_list.append(neighbor)
    return None

# Example
grid = [
    [0, 0, 1, 0],
    [1, 0, 1, 0],
    [0, 0, 0, 0]
]
start = (0, 0)
goal = (2, 3)
print("A* Path:", a_star(grid, start, goal))
```

```python
def a_star(grid, start, goal):
    from collections import deque

    def heuristic(a, b):
        return abs(a[0] - b[0]) + abs(a[1] - b[1])

    dirs = [(-1,0), (1,0), (0,-1), (0,1)]
    open_list = [start]
    came = {}
    g = {start: 0}

    while open_list:
        curr = min(open_list, key=lambda x: g[x] + heuristic(x, goal))
        if curr == goal:
            path = deque()
            while curr in came:
                path.appendleft(curr)
                curr = came[curr]
            path.appendleft(start)
            return list(path)
        open_list.remove(curr)
        for dx, dy in dirs:
            nb = (curr[0]+dx, curr[1]+dy)
            if 0<=nb[0]<len(grid) and 0<=nb[1]<len(grid[0]) and grid[nb[0]][nb[1]]==0:
                tentative_g = g[curr] + 1
                if tentative_g < g.get(nb, float('inf')):
                    g[nb] = tentative_g
                    came[nb] = curr
                    if nb not in open_list:
                        open_list.append(nb)
    return None

rows, cols = map(int, input("Rows Columns: ").split())
grid = [list(map(int, input().split())) for _ in range(rows)]
sx, sy = map(int, input("Start row col: ").split())
gx, gy = map(int, input("Goal row col: ").split())
start, goal = (sx, sy), (gx, gy)

path = a_star(grid, start, goal)
print("A* Path:", path if path else "No path found.")

if path:
    for i in range(rows):
        print(' '.join(
            'S' if (i,j)==start else
            'G' if (i,j)==goal else
            '*' if (i,j) in path and (i,j)!=start and (i,j)!=goal else
            '█' if grid[i][j]==1 else '.' for j in range(cols)
        ))
```

```python
def a_star(grid, start, goal):   # Create a function
    H = lambda a, b: abs(a[0]-b[0]) + abs(a[1]-b[1]) # Heuristic function Manhattan dis.
    OPEN, CAME, G = [start], {}, {start:0} #Initialize data str.
    while OPEN:
        curr = min(OPEN, key=lambda x: G[x] + H(x, goal))  #select best node
        OPEN.remove(curr) #remove current node
        
        if curr == goal: # checck reach goal
            path = []    #reconstract path when goal is found
            while curr in CAME: path.append(curr); curr = CAME[curr]
            return [start]+path[::-1]
            
        for dx,dy in [(-1,0),(1,0),(0,-1),(0,1)]:  #Check all four directions
            nb = (curr[0]+dx, curr[1]+dy)    #calculate neighbor position

            #Check if neighbor is valid
            if 0<=nb[0]<len(grid) and 0<=nb[1]<len(grid[0]) and grid[nb[0]][nb[1]]==0: 
                #Check if this path to neighbor is better
                if G[curr]+1 < G.get(nb,99):
                    G[nb]=G[curr]+1; CAME[nb]=curr; OPEN+=[nb] #Update neighbor if better path found
    return None

# Input Example: All values are prompted one by one
r = int(input("Rows: ")); c = int(input("Columns: "))
grid = [list(map(int, input().split())) for _ in range(r)]
sx, sy = map(int, input("Start (row col): ").split())
gx, gy = map(int, input("Goal (row col): ").split())
path = a_star(grid, (sx,sy), (gx,gy))
print("Path:", path if path else "No path")

if path:
    for i in range(r):
        print(' '.join(
            'S' if (i,j)==(sx,sy) else
            'G' if (i,j)==(gx,gy) else
            '*' if (i,j) in path else
            '█' if grid[i][j] else '.' for j in range(c)
        ))


for i in range(r):
    for j in range(c):
        if (i, j) == (sx, sy):
            print('S', end=' ')
        elif (i, j) == (gx, gy):
            print('G', end=' ')
        elif (i, j) in path:
            print('*', end=' ')
        elif grid[i][j] == 1:
            print('\u2588', end=' ')  # or print('█', end=' ')
        else:
            print('.', end=' ')
    print()
```
